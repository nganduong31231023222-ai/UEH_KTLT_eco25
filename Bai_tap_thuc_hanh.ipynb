```python
# Bài 1: In nhãn InNhan

WIDTH = 26  # tổng số ký tự của mỗi dòng (số * trong viền)
INNER = WIDTH - 4  # phần nội dung giữa "* " và " *"

def line(text=""):
    return f"* {text:<{INNER}} *"

print("*" * WIDTH)
print(line("Truong: Dai Hoc UEH"))
print(line("Khoa: CNTT"))
print(line("Ho ten: "))
print("*" * WIDTH)
```

```python
# Bài 2: Tính diện tích và chu vi hình tròn HinhTron

# Nhập bán kính R
R = float(input("Nhap ban kinh R: "))

# Tính diện tích và chu vi
S = 3.14 * R * R
P = 2 * 3.14 * R

# In kết quả với 1 số lẻ thập phân
print(f"Dien tich S = {S:.1f}")
print(f"Chu vi P = {P:.1f}")
```

```python
# Bài 3: Đổi sang giây DoiSangGiay

# Nhập thời gian
h = int(input("Nhap so gio: "))
m = int(input("Nhap so phut: "))
s = int(input("Nhap so giay: "))

# Đổi sang giây
tong_giay = h * 3600 + m * 60 + s

# In kết quả
print(f"Tong so giay cua {h}:{m}:{s} la {tong_giay} giay")
```

```python
# Bài 4: Đổi sang giờ phút giây DoiSangGioPhutGiay

# Nhập tổng số giây
t = int(input("Nhap vao tong so giay: "))

# Tính giờ, phút, giây
h = t // 3600
m = (t % 3600) // 60
s = t % 60

# In kết quả
print(f"{t} giay co dang {h}:{m}:{s}")
```

```python
# Bài 5: Tính diện tích tam giác DienTichTamGiac

import math

# Nhập độ dài 3 cạnh
a = float(input("Nhap do dai canh a: "))
b = float(input("Nhap do dai canh b: "))
c = float(input("Nhap do dai canh c: "))

# Kiểm tra điều kiện tam giác
if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2  # nửa chu vi
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))  # công thức Heron
    print(f"Dien tich tam giac S = {S:.2f}")
else:
    print("Ba cạnh không tạo thành một tam giác hợp lệ.")
```

```python
# Bài 6: Hàm Mũ TinhMu

# Nhập dữ liệu
a = float(input("Nhap so thuc a: "))
n = int(input("Nhap so mu n: "))

# Tính a^n
ket_qua = a ** n

# In kết quả
print(f"Ket qua {a}^{n} = {ket_qua}")
```

```python
# Bài 7: Đảo số DaoSo

# Nhập số nguyên 4 chữ số
x = int(input("Nhap so nguyen x (co 4 chu so): "))

# Cách 1: Dùng chuỗi
y = int(str(x)[::-1])

# In kết quả
print(f"So dao cua {x} la {y}")
```

```python
# Bài 8: Độ sang Radian DoSangRadian
import math

# Nhập góc theo độ
degree = float(input("Nhap goc (do): "))

# Đổi sang radian
radian = degree * math.pi / 180

# In kết quả với 4 chữ số thập phân
print(f"{degree} do = {radian:.4f} radian")
```

```python
# Bài 9: Hàm số 1 (HamSo1.py)
import math

# Nhập giá trị x
x = float(input("Nhap x: "))

# Tính y1 và y2
y1 = 4 * (x**2 + 10 * math.sqrt(x) + 3*x + 1)
y2 = (math.sin(math.pi * x**2) + math.sqrt(x**2 + 1)) / (math.exp(2*x) + math.cos((math.pi/4) * x))

# In kết quả với 2 chữ số thập phân
print(f"y1 = {y1:.2f}")
print(f"y2 = {y2:.2f}")
```

```python
# Bài 10: Khoảng cách (KhoangCach.py)
import math

# Nhập tọa độ A và B
x1 = float(input("Nhap x1: "))
y1 = float(input("Nhap y1: "))
x2 = float(input("Nhap x2: "))
y2 = float(input("Nhap y2: "))

# Tính khoảng cách
d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# In kết quả với 2 chữ số thập phân
print(f"Khoang cach giua A({x1},{y1}) va B({x2},{y2}) la {d:.2f}")
```

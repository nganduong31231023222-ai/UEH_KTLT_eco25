```python
# ham kiem tra so nguyen to
def ktsnt(n):
    if n < 2:  # 0, 1 khong la so nguyen to
        return False
    for i in range(2, int(n**(0.5)) + 1):  # can bac hai cua n
        if (n % i == 0):
            return False
    return True


# nhap vao so nguyen can kiem tra
a = int(input("Nhap vao so nguyen: "))

if ktsnt(a):
    print("%d la so nguyen to" % a)
else:
    print("%d khong la so nguyen to" % a)
```

```python
import math

n = int(input("Nhập số nguyên dương n (>0): "))

# 1. Kiểm tra đối xứng
if str(n) == str(n)[::-1]:
    print("→ Số đối xứng")
else:
    print("→ Không phải số đối xứng")

# 2. Kiểm tra chính phương
if int(math.sqrt(n))**2 == n:
    print("→ Số chính phương")
else:
    print("→ Không phải số chính phương")

# 3. Kiểm tra nguyên tố
if n > 1 and all(n % i != 0 for i in range(2, int(math.sqrt(n))+1)):
    print("→ Số nguyên tố")
else:
    print("→ Không phải số nguyên tố")

# 4. Chữ số lớn nhất và nhỏ nhất
digits = [int(ch) for ch in str(n)]
print("→ Chữ số lớn nhất:", max(digits))
print("→ Chữ số nhỏ nhất:", min(digits))

# 5. Kiểm tra tăng dần / giảm dần
if digits == sorted(digits):
    print("→ Các chữ số tăng dần")
elif digits == sorted(digits, reverse=True):
    print("→ Các chữ số giảm dần")
else:
    print("→ Không tăng/giảm dần")
```

```python
n = int(input("Nhập n: "))

# a. 1 + 2 + … + n
s1 = sum(range(1, n+1))

# b. 1² + 2² + … + n²
s2 = sum(i**2 for i in range(1, n+1))

# c. 1 + 1/2 + … + 1/n
s3 = sum(1/i for i in range(1, n+1))

# d. n! (giai thừa)
fact = 1
for i in range(1, n+1):
    fact *= i

# e. 1! + 2! + … + n!
s5 = 0
f = 1
for i in range(1, n+1):
    f *= i
    s5 += f

print("a.", s1)
print("b.", s2)
print("c.", s3)
print("d.", fact)
print("e.", s5)
```

```python
import math

# Hàm kiểm tra số nguyên tố
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return False
    return True

# Nhập số dòng và số cột
n, m = map(int, input().split())

A = []
for _ in range(n):
    row = list(map(int, input().split()))
    A.append(row)

# Đếm số nguyên tố trong từng cột
prime_count = [0] * m
for j in range(m):
    for i in range(n):
        if is_prime(A[i][j]):
            prime_count[j] += 1

# Tìm số lượng nguyên tố lớn nhất
max_count = max(prime_count)

# Lấy tất cả cột có số nguyên tố = max_count
cols = [str(j) for j in range(m) if prime_count[j] == max_count]

print("Cac cot nhieu nguyen to nhat:", " ".join(cols))
```

```python
def input_square_matrix(n):
    matrix = []
    for i in range(n):
        row = list(map(int, input().split()))
        matrix.append(row)
    return matrix

def print_matrix(matrix):
    for row in matrix:
        print(row)

def count_upper_triangle_numbers(matrix):
    """
    Đếm các loại số trong tam giác trên của đường chéo chính
    (không bao gồm các phần tử trên đường chéo chính)
    """
    n = len(matrix)
    negative_count = 0
    positive_count = 0
    zero_count = 0

    # Duyệt qua tam giác trên của đường chéo chính
    for i in range(n):
        for j in range(i + 1, n):  # j > i để chỉ lấy tam giác trên
            element = matrix[i][j]
            if element < 0:
                negative_count += 1
            elif element > 0:
                positive_count += 1
            else:
                zero_count += 1

    return negative_count, positive_count, zero_count

def count_upper_triangle_anti_diagonal(matrix):
    """
    Đếm các loại số trong tam giác trên của đường chéo phụ
    (không bao gồm các phần tử trên đường chéo phụ)
    """
    n = len(matrix)
    negative_count = 0
    positive_count = 0
    zero_count = 0

    # Duyệt qua tam giác trên của đường chéo phụ
    # Đường chéo phụ: i + j = n - 1
    # Tam giác trên: i + j < n - 1
    for i in range(n):
        for j in range(n):
            if i + j < n - 1:  # Tam giác trên đường chéo phụ
                element = matrix[i][j]
                if element < 0:
                    negative_count += 1
                elif element > 0:
                    positive_count += 1
                else:
                    zero_count += 1

    return negative_count, positive_count, zero_count

def main():
    n = int(input())
    matrix = input_square_matrix(n)

    # Đếm các loại số trong tam giác trên đường chéo chính
    neg_main, pos_main, zero_main = count_upper_triangle_numbers(matrix)

    # Đếm các loại số trong tam giác trên đường chéo phụ
    neg_anti, pos_anti, zero_anti = count_upper_triangle_anti_diagonal(matrix)

    # Hiển thị kết quả cho đường chéo chính
    print("Trong nua tam giac tren cheo chinh:")
    print(f"+ {neg_main} so am")
    print(f"+ {pos_main} so duong")
    print(f"+ {zero_main} so khong")

    # Hiển thị kết quả cho đường chéo phụ
    print("Trong nua tam giac tren cheo phu:")
    print(f"+ {neg_anti} so am")
    print(f"+ {pos_anti} so duong")
    print(f"+ {zero_anti} so khong")

main()
```

```python
def input_square_matrix(n):
    """Nhập ma trận vuông cấp n"""
    matrix = []
    for _ in range(n):
        row = list(map(int, input().split()))
        matrix.append(row)
    return matrix

def parallel_diagonals_main(matrix):
    """
    Liệt kê các đường song song với đường chéo chính
    Trả về list các list
    """
    n = len(matrix)
    diagonals = []

    # Các đường song song với đường chéo chính có cùng i-j = k
    # Giá trị k chạy từ -(n-1) đến n-1
    for k in range(-(n-1), n):
        diagonal = []
        for i in range(n):
            j = i - k
            if 0 <= j < n:
                diagonal.append(matrix[i][j])
        diagonals.append(diagonal)
    return diagonals

def parallel_diagonals_anti(matrix):
    """
    Liệt kê các đường song song với đường chéo phụ
    Trả về list các list
    """
    n = len(matrix)
    diagonals = []

    # Các đường song song với đường chéo phụ có cùng i+j = k
    # Giá trị k chạy từ 0 đến 2n-2
    for k in range(0, 2*n - 1):
        diagonal = []
        for i in range(n):
            j = k - i
            if 0 <= j < n:
                diagonal.append(matrix[i][j])
        diagonals.append(diagonal)
    return diagonals

def main():
    n = int(input())
    matrix = input_square_matrix(n)

    main_diagonals = parallel_diagonals_main(matrix)
    anti_diagonals = parallel_diagonals_anti(matrix)

    print("Cac duong song song cheo chinh:", main_diagonals)
    print("Cac duong song song cheo phu:", anti_diagonals)

main()
```

```python
# Nhập ma trận vuông
n = int(input())
A = []
for _ in range(n):
    row = list(map(int, input().split()))
    A.append(row)

# Liệt kê các dòng theo yêu cầu
dong_list = []
for i in range(n):
    if i % 2 == 0:          # hàng chẵn (0-indexed) -> left to right
        dong_list.append(A[i])
    else:                   # hàng lẻ -> right to left
        dong_list.append(A[i][::-1])

print("Cac dong:", end=" ")
for row in dong_list:
    print(row, end="")
print()

# Liệt kê các cột theo yêu cầu
cot_list = []
for j in range(n):
    col = []
    if j % 2 == 0:          # cột chẵn -> top to bottom
        for i in range(n):
            col.append(A[i][j])
    else:                   # cột lẻ -> bottom to top
        for i in range(n-1, -1, -1):
            col.append(A[i][j])
    cot_list.append(col)

print("Cac cot:", end=" ")
for col in cot_list:
    print(col, end="")
print()
```

```python
import re

def solve_word_histogram():
    # Đọc số lượng câu
    n = int(input())

    # Dictionary để lưu tần số và thứ tự xuất hiện đầu tiên
    word_count = {}
    word_order = {}
    order_counter = 0

    # Đọc n câu
    for _ in range(n):
        sentence = input().strip()

        # Tách từ bằng các dấu phân cách: khoảng trắng, dấu phẩy, chấm phẩy, dấu ngang, dấu hỏi
        words = re.split(r'[ ,;?-]+', sentence)

        for word in words:
            # Loại bỏ khoảng trắng thừa và chuyển về chữ thường
            word = word.strip().lower()

            # Bỏ qua từ rỗng
            if not word:
                continue

            # Đếm tần số
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1
                # Lưu thứ tự xuất hiện đầu tiên
                word_order[word] = order_counter
                order_counter += 1

    # Sắp xếp theo thứ tự xuất hiện đầu tiên, sau đó theo thứ tự chữ cái
    sorted_words = sorted(word_count.keys(), key=lambda x: (word_order[x], x))

    # In kết quả
    for word in sorted_words:
        print(f"{word}: {word_count[word]}")

if __name__ == "__main__":
    solve_word_histogram()
```

```python
def reverse_words_in_lines():
    # Nhập số lượng chuỗi
    n = int(input())

    # Xử lý từng chuỗi
    for _ in range(n):
        line = input().strip()
        # Tách từ theo khoảng trắng, đảo thứ tự, rồi nối lại
        reversed_line = ' '.join(line.split()[::-1])
        print(reversed_line)

reverse_words_in_lines()
```

```python
def majority_element(nums):
    """
    Tìm phần tử xuất hiện hơn n/2 lần trong nums
    (phần tử quan trọng luôn tồn tại)
    Dùng Boyer-Moore Voting Algorithm O(n) + O(1)
    """
    candidate = None
    count = 0
    for num in nums:
        if count == 0:
            candidate = num
            count = 1
        elif num == candidate:
            count += 1
        else:
            count -= 1
    return candidate

def main():
    t = int(input())  # số lượng test case
    for _ in range(t):
        nums = list(map(int, input().split()))
        print(majority_element(nums))

main()
```

```python
def is_happy_trace(n: int) -> bool:
    seen = set()
    while n != 1:
        if n in seen:
            return False
        seen.add(n)
        digits = [int(d) for d in str(n)]
        print("+".join(f"{d}^2" for d in digits), "=", sum(d**2 for d in digits), "→", end=" ")
        n = sum(d**2 for d in digits)
    print("1")  # kết thúc tại 1
    return True

n = int(input())
print(is_happy_trace(n))
```

```python
def is_isomorphic(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    # Dictionary để ánh xạ s -> t
    mapping_s_t = {}
    # Set để đảm bảo 1 ký tự t chỉ được ánh xạ 1 lần
    mapped_chars = set()

    for cs, ct in zip(s, t):
        if cs in mapping_s_t:
            # Nếu đã ánh xạ trước đó, phải khớp với ct hiện tại
            if mapping_s_t[cs] != ct:
                return False
        else:
            # Nếu ct đã được ánh xạ từ ký tự khác → False
            if ct in mapped_chars:
                return False
            mapping_s_t[cs] = ct
            mapped_chars.add(ct)
    return True

# Ví dụ sử dụng
pairs = [
    ("egg", "add"),
    ("foo", "bar"),
    ("paper", "title")
]

for s, t in pairs:
    print(f"s = {s}, t = {t} -> {is_isomorphic(s, t)}")
```

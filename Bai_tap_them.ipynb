```python
# Input
a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())

# Process
def maxmin5(a, b, c, d, e):
    values = [a, b, c, d, e]
    vmax = max(values)
    vmin = min(values)
    return vmin, vmax

vmin, vmax = maxmin5(a, b, c, d, e)

# Output
print("%d %d" % (vmin, vmax))
```

```python
import math

# Input
a = float(input())
b = float(input())
c = float(input())

# Process
def solvQuadEq(a, b, c):
    if a == 0:  # bậc 1 hoặc vô nghiệm
        if b == 0:
            if c == 0:
                return "Unlimited solutions"
            else:
                return "No solution"
        else:
            x = -c / b
            return "x1=%.2f" % x
    else:  # bậc 2
        delta = b*b - 4*a*c
        if delta < 0:
            return "No solution"
        elif delta == 0:
            x = -b / (2*a)
            return "x1=%.2f" % x
        else:
            x1 = (-b - math.sqrt(delta)) / (2*a)
            x2 = (-b + math.sqrt(delta)) / (2*a)
            return "x1=%.2f, x2=%.2f" % (x1, x2)

# Output
print(solvQuadEq(a, b, c))
```

```python
# Input
m = int(input())
y = int(input())

# Process
def is_leap(year):
    return (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

def total_day_month(m, y):
    if m in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    elif m in [4, 6, 9, 11]:
        return 30
    elif m == 2:
        if is_leap(y):
            return 29
        else:
            return 28
    else:
        return -1  # tháng không hợp lệ

# Output
print(total_day_month(m, y))
```

```python
# Input
n = int(input())

# Process
d1 = n // 1000         # chữ số hàng nghìn
d2 = (n // 100) % 10   # chữ số hàng trăm
d3 = (n // 10) % 10    # chữ số hàng chục
d4 = n % 10            # chữ số hàng đơn vị

ans = max(d1, d2, d3, d4)

# Output
print(ans)
```

```python
# Input
math = float(input())     # điểm Toán
physic = float(input())   # điểm Lý
chemistry = float(input())# điểm Hóa

# Process
avg = (math * 2 + physic * 3 + chemistry) / 6

if 8 <= avg <= 10:
    rating = "Good"
elif 6.5 <= avg < 8:
    rating = "Pretty"
elif 5 <= avg < 6.5:
    rating = "Average"
else:
    rating = "Weak"

# Output
print("Average point = %.2f, rating %s." % (avg, rating))
```

```python
# Input
old_index = int(input())   # Chỉ số nước cũ
new_index = int(input())   # Chỉ số nước mới
people = int(input())      # Số người

# Process
consumed = new_index - old_index   # Số m3 đã dùng
quota1 = people * 4                # Mức 1: 4m3/người
quota2 = people * 2                # Mức 2: 2m3/người

money = 0

if consumed <= quota1:
    money = consumed * 4400
elif consumed <= quota1 + quota2:
    money = quota1 * 4400 + (consumed - quota1) * 8300
else:
    money = quota1 * 4400 + quota2 * 8300 + (consumed - quota1 - quota2) * 10500

# Output
print("Payment for %d m^3 in month is %s Đ." % (consumed, format(money, ",").replace(",", ".")))
```

```python
# Input
d = int(input())
m = int(input())
y = int(input())

# Process
def isleap(year):
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

def days_in_month(month, year):
    if month in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    elif month in [4, 6, 9, 11]:
        return 30
    else:  # tháng 2
        return 29 if isleap(year) else 28

# ngoại lệ theo đề
if d == 31 and m == 1 and y == 2015:
    d = 1
    m = 2
    y = 2016
else:
    d += 1
    if d > days_in_month(m, y):
        d = 1
        m += 1
        if m > 12:
            m = 1
            y += 1

# Output
print(f"{d}/{m}/{y}")
```

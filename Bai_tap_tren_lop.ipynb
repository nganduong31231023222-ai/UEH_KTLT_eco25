```python
# Input a, b, c
a = int(input("Nhap a: "))
b = int(input("Nhap b: "))
c = int(input("Nhap c: "))

# max a, b, c
def max_fn(a, b):
    return a if a > b else b
# end max_fn

vmax = max_fn(a, b)
vmax = max_fn(vmax, c)

# output
print("Max(%d, %d, %d) = %d" % (a, b, c, vmax))
```

```python
# Input a, b với phương trình ax + b = 0
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))

# Process
def solv1deg(a, b):
    """
    Giải phương trình bậc 1: ax + b = 0
    Trả về:
        nsol = -1  -> vô số nghiệm
        nsol = 0   -> vô nghiệm
        nsol = 1   -> có 1 nghiệm
        x = nghiệm (nếu có)
    """
    nsol, x = None, None
    if a == 0:
        if b == 0:
            nsol = -1   # vô số nghiệm
        else:
            nsol = 0    # vô nghiệm
    else:
        nsol = 1        # có 1 nghiệm
        x = -b / a
    return nsol, x
# end solv1deg

# Gọi hàm
nsol, x = solv1deg(a, b)

# Output
if nsol == -1:
    print("%.2fx + %.2f = 0: unlimited solutions" % (a, b))
elif nsol == 0:
    print("%.2fx + %.2f = 0: no solution" % (a, b))
else:
    print("%.2fx + %.2f = 0: x = %.2f" % (a, b, x))
```

```python
# Input
a = float(input("Nhập số thứ nhất: "))
b = float(input("Nhập số thứ hai: "))
op = input("Nhập phép toán (+, -, *, /): ")

# Process
def calculate(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        if b == 0:
            return None
        else:
            return a / b
    else:
        return "Phép toán không hợp lệ"
# end calculate

result = calculate(a, b, op)

# Output
if result is None:
    print(f"{a:.2f} {op} {b:.2f} = Không tính được (chia cho 0)")
elif isinstance(result, str):
    print(result)
else:
    print(f"{a:.2f} {op} {b:.2f} = {result:.2f}")
```

```python
# Input
old_kwh = int(input("Nhập chỉ số cũ: "))
new_kwh = int(input("Nhập chỉ số mới: "))

# Số điện tiêu thụ
consumed = new_kwh - old_kwh

# Process
def calc_electric(consumed):
    price = 0
    if consumed <= 100:
        price = consumed * 1242
    elif consumed <= 150:
        price = 100 * 1242 + (consumed - 100) * 1304
    elif consumed <= 200:
        price = 100 * 1242 + 50 * 1304 + (consumed - 150) * 1651
    elif consumed <= 300:
        price = 100 * 1242 + 50 * 1304 + 50 * 1651 + (consumed - 200) * 1788
    elif consumed <= 400:
        price = 100 * 1242 + 50 * 1304 + 50 * 1651 + 100 * 1788 + (consumed - 300) * 1912
    else:
        price = (100 * 1242 + 50 * 1304 + 50 * 1651 +
                 100 * 1788 + 100 * 1912 + (consumed - 400) * 1962)

    # VAT 10%
    total = int(price * 1.1)
    return total
# end calc_electric

amount = calc_electric(consumed)

# Output
print(f"The amount to pay for {consumed} kWh consumed in the month is {amount:,} VND.")
```

```python
# Input
d = int(input("Nhập ngày: "))
m = int(input("Nhập tháng: "))
y = int(input("Nhập năm: "))

# Hàm kiểm tra năm nhuận
def isleap(y):
    return (y % 400 == 0) or (y % 4 == 0 and y % 100 != 0)

# Hàm số ngày trong tháng
def days_in_month(m, y):
    if m in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    elif m in [4, 6, 9, 11]:
        return 30
    elif m == 2:
        return 29 if isleap(y) else 28
    else:
        return 0  # Tháng không hợp lệ

# Hàm tính ngày trước
def previous_day(d, m, y):
    if d == 1 and m == 1:  # Trường hợp đầu năm
        return 31, 12, y - 1
    elif d == 1:  # Trường hợp đầu tháng
        pm = m - 1
        return days_in_month(pm, y), pm, y
    else:  # Ngày bình thường
        return d - 1, m, y

# Process
pd, pm, py = previous_day(d, m, y)

# Output
print(f"Previous day of {d}/{m}/{y} is {pd}/{pm}/{py}.")
```

```python
# Input
a = int(input())
b = int(input())
c = int(input())

# Process
def check_triangle(a, b, c):
    # Sắp xếp các cạnh
    x, y, z = sorted([a, b, c])
    if x + y > z:
        return True, z  # z là cạnh lớn nhất
    else:
        return False, z

is_triangle, vmax = check_triangle(a, b, c)

# Output
print("Max(%d, %d, %d) = %d." % (a, b, c, vmax))
if is_triangle:
    print("It is a valid triangle.")
else:
    print("It is NOT a valid triangle.")
```

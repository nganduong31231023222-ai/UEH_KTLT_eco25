```python
amount = 100
interest = 10
years = 7

future_value = amount * ((1 + 0.01 * interest)**years)
print(round(future_value, 2))
```

```python
# Python Function
def getfv(pv, i, n):
    fv = pv * ((1 + i) ** n)
    return fv

fv = getfv(100, 0.1, 7)
print(round(fv, 2))
```
